```python
import numpy as np

test_scores = np.array([55, 80, 45, 90, 60, 70, 30])
```

## Exercise 1 — Combine three conditions

Write a function that returns only the scores that are between 50 and 85, **inclusive on both ends**.

```python
def mid_range_scores(scores):
    """Return only scores between 50 and 85, inclusive on both ends."""
    mask = (scores > 50) & (scores < 85)
    return scores[mask] 
```

```python
# Test cell
result = mid_range_scores(test_scores)
print(result)
assert list(result) == [55, 80, 60, 70], f"Expected [55, 80, 60, 70], got {list(result)}"
print("Passed!")
```

## Exercise 2 — Label with three categories using `np.where()`

`np.where()` can be nested to handle more than two outcomes.

Write a function that labels each score:
- `'Fail'` if below 50
- `'Pass'` if between 50 and 74 (inclusive)
- `'Distinction'` if 75 or above

Hint: nest a second `np.where()` inside the `x` or `y` argument of the first one.

```python
def label_scores(scores):
    """Label each score as 'Fail', 'Pass', or 'Distinction'."""
    return np.where(scores > 50, np.where((scores > 50) & (74 > scores), "Pass", "Distinction"), "Fail") 
```

```python
# Test cell
result2 = label_scores(test_scores)
print(result2)
expected2 = ['Pass', 'Distinction', 'Fail', 'Distinction', 'Pass', 'Pass', 'Fail']
assert list(result2) == expected2, f"Expected {expected2}, got {list(result2)}"
print("Passed!")
```

## Exercise 3 — Mask one array using another

So far the mask has come from the same array you're filtering. Now use a mask built from *one* array to filter a *different* array.

You're given `names` and `test_scores` (same length, matching by position). Write a function that returns the names of everyone who scored 60 or above.

```python
names = np.array(['Amir', 'Bella', 'Carl', 'Dana', 'Eli', 'Farah', 'Gus'])

def high_scorers(names, scores, threshold):
    """Return the names of everyone whose score is >= threshold."""
    mask = scores >= threshold
    return names[mask]
```

```python
# Test cell
result3 = high_scorers(names, test_scores, 60)
print(result3)
expected3 = ['Bella', 'Dana', 'Eli', 'Farah']
assert list(result3) == expected3, f"Expected {expected3}, got {list(result3)}"
print("Passed!")
```
