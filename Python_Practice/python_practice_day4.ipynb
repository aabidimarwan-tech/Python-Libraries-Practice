## Setup

Run this first to create the sample DataFrame we'll use throughout.

```python
import pandas as pd

df = pd.DataFrame({
    "name": ["Ali", "Sara", "Omar", "Lina", "Yousra", "Karim", "Nadia", "Bilal", "Mima"],
    "grade": ["A", "B", "A", "C", "B", "A", "C", "B", "A"],
    "score": [92, 78, 65, 55, 88, 95, 40, 60, 67]
})

```

## 1. Combining conditions with `&`, `|`, `~`

In plain Python you'd write `and` / `or`, but pandas needs `&` / `|` / `~` instead — **and each condition needs its own parentheses**.

Why the parentheses are required: Python's operator precedence would otherwise try to evaluate `&` before `>=`, breaking the expression. So writing it without parentheses will error or misbehave — always wrap each condition.

```python
# AND: score >= 60 AND grade == "A"
df[(df["score"] >= 60) & (df["grade"] == "A")]


```

```python
# OR: grade == "A" OR grade == "B"
df[(df["grade"] == "A") | (df["grade"] == "B")]
```

```python
# NOT: students NOT in grade "A" 
df[~(df["grade"] == "A")] 
```

## 2. Sorting with `.sort_values()`

```python
# Ascending (default)
df.sort_values("score") 
```

```python
# Descending
df.sort_values("score", ascending=False)
```

```python
# Sort by grade first, then score within each grade
df.sort_values(["grade", "score"])
```

## Combining filter + sort in one line

```python
# Filter passing students, then show highest-to-lowest
df[df["score"] >= 60].sort_values("score", ascending=False)
```
