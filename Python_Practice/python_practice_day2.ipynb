## Exercise 2: Nested dictionaries
Given an inventory of items with price and quantity:
- `total_value(inventory)` → sum of price*qty across all items
- `most_valuable_item(inventory)` → name of the item with the highest price*qty
- Print each item as: `"apples: $0.50 x 100 = $50.00"`

```python
inventory = {
    "apples": {"price": 0.5, "qty": 100},
    "bananas": {"price": 0.3, "qty": 150},
    "cherries": {"price": 3.0, "qty": 20}
}

def total_value(inv):
    """Return the sum of price*qty across all items."""
    totals  = {}
    for item in inv:
        price = inv[item]["price"]
        qty = inv[item]["qty"] 

        totals[item] = price * qty

    print(sum(totals.values()))  

    return totals


def most_valuable_item(inv):
    """Return the name of the item with the highest price*qty."""
    totals = total_value(inv)
    return max(totals, key = totals.get)



def print_inventory(inv):
    """Print each item as: 'apples: $0.50 x 100 = $50.00"""
    for item in inv:
        price = inv[item]["price"]
        qty = inv[item]["qty"]
        total = price * qty
        print(f" {item} : ${price} x {qty} = {total} ") 





    
```

```python
print(total_value(inventory))
print(most_valuable_item(inventory))
print_inventory(inventory)
```

## Exercise 3: Error handling
Use `try`/`except` — not `if` checks — to handle these safely.

```python

def safe_divide(a, b):
    """Return a/b, or None (with a printed message) if b is 0."""
    try:
        return a/b
    except ZeroDivisionError:
        print("Cannot devide by Zero")
        return None  




def parse_int(s):
    """Try to convert s to int. Return None if it fails (e.g. s = 'abc')."""
    try:
        return int(s)
    except ValueError:
        print("the value is not a number")
        return None 

```

```python
print(safe_divide(10, 0))
print(safe_divide(10, 2))
print(parse_int("42"))
print(parse_int("abc"))
```

## Exercise 4: File I/O
Basic file handling — the bridge to pandas.
- Write `scores.txt` yourself first, one line per student: `name,score` (at least 5 students)
- `read_scores(filename)` → reads the file, returns `{name: int(score)}`
- `write_scores(filename, scores_dict)` → writes a dict back to a file in the same format

```python
def read_scores(filename):
    """Read the file, return dict {name: int(score)}."""
    scores = {} 
    with open( filename, "r") as f:
        for line in f:
            line = line.strip()
            if not line:
                continue 
            name, score = line.split(",")
            scores[name] = int(score) 

    return scores 





    

def write_scores(filename, scores_dict):
    """Write a dict of {name: score} back to a file, same format."""
    with open(filename, "w") as f: 
        for name in scores_dict:
            score = scores_dict[name]
            line = f" {name}, {score} \n"
            f.write(line)
        



    
```
