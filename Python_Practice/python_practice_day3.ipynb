## Exercise 1: Loading a DataFrame
- `load_scores_df(filename)` → reads a `name,score` file into a pandas DataFrame with columns `["name", "score"]`. Make sure the `score` column is an integer type, not a string.

```python
import pandas as pd

def load_scores_df(filename):
    """Read a 'name,score' file into a DataFrame with columns ['name', 'score']."""

    df = {"name": [], "score": []}
    with open(filename, "r") as f:
        for line in f:
            line = line.strip()
            if not line:
                continue
            name, score = line.split(",")
            df["name"].append(name)
            df["score"].append(int(score))

    StudentData = pd.DataFrame(df) 
    return StudentData






    # stu_data = []
    # with open(filename, "r") as f:
    #     for line in f:
    #         line = line.strip()
    #         if not line:
    #             continue
    #         name, score = line.split(",")
    #         stu_data.append((name, int(score)))
    # Student_Data = pd.DataFrame(stu_data, columns = ["name", "score"])
    # return Student_Data 
  
```

## Exercise 2: Exploring the DataFrame
- `top_n(df, n)` → returns the top `n` students by score, as a DataFrame, sorted highest to lowest
- `average_score(df)` → returns the average score as a float
- `score_range(df)` → returns a tuple `(min_score, max_score)`

```python
def top_n(df, n):
    """Return the top n students by score, sorted highest to lowest."""
    sorted_df = df.sort_values(by = "score", ascending = False)
    return sorted_df.head(n) 


def average_score(df):
    """Return the average score as a float."""
    average_df = df["score"].mean()
    return f"{average_df:.2f}" 


    # avearge = df["score"].sum()/df["score"].count 


def score_range(df):
    """Return a tuple (min_score, max_score)."""
    min_score = df["score"].min()
    max_score = df["score"].max()

    return (int(min_score), int(max_score))



```

```python
print(top_n(df, 3))
print(average_score(df))
print(score_range(df))
```

## Exercise 3: Filtering and adding columns
- `passing_students(df, threshold)` → returns only the rows where `score >= threshold`
- `add_grade_column(df)` → adds a new column `grade` to the DataFrame: `"A"` if score >= 90, `"B"` if >= 80, `"C"` if >= 70, else `"F"`

```python
def passing_students(df, threshold):
    """Return only the rows where score >= threshold."""
    passed_stu = df[df["score"] >= threshold]
    return passed_stu 



# first method: 

def grade_by_score(df):
    if df["score"] >= 90:
        return "A"
    elif df["score"] >= 80:
        return "B"
    elif df["score"] >= 70:
        return "C"
    else:
        return "F"


def add_grade_column(df):
    """Add a 'grade' column: A >= 90, B >= 80, C >= 70, else F."""
    Grades = grade_by_score

    df["grade"] = df.apply(Grades, axis = 1) 
    return df 



# # second method:

# def grade_by_score(score):
#     if   score >= 90:
#         return "A"
#     elif score >= 80:
#         return "B"
#     elif score >= 70:
#         return "C"
#     else:
#         return "F"


# def add_grade_column(df):
#     """Add a 'grade' column: A >= 90, B >= 80, C >= 70, else F."""
#     Grades = grade_by_score

#     df["grade"] = df["score"].apply(Grades) 
#     return df 






```

## Exercise 4: Grouping
- `count_by_grade(df)` → returns a Series (or DataFrame) showing how many students got each grade
- `average_by_grade(df)` → returns the average score for each grade group

```python
def count_by_grade(df):
    """Return the count of students per grade.""" 
    student_per_grade = df.groupby("grade")
    count_grade = student_per_grade.size() 
    return count_grade 

     


def average_by_grade(df):
    """Return the average score per grade."""
    student_groups = df.groupby("grade")
    average = student_groups["score"].mean()
    return average  

```

```python
print(count_by_grade(df))
print(average_by_grade(df))
```

## Exercise 5: Saving results
- `save_top_students(df, filename, n)` → writes the top `n` students (name and score only) to a new CSV file using pandas (`.to_csv`), not manual file writing

```python
def save_top_students(df, filename, n):
    """Write the top n students (name, score only) to a CSV file using pandas."""
    sorted_student = df.sort_values(by = "score", ascending = False)
    top_student = sorted_student.head(n)

    top_student.to_csv(filename, index = False)

    
```
