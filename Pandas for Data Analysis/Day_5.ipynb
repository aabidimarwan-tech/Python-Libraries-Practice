### Ordering in DataFrames


```python
import pandas as pd
```

```python
df = pd.read_csv(r"C:\Users\seid\Downloads\filtering-using-sql-ggyou.zip", encoding = "cp1252")
```

```python
df[df["Company"] == "Razer"].sort_values(by = ["Inches","Price_usd"], ascending = [False,True]) 
```

```python
df[["Company", "Inches", "Price_usd"]][df["Company"] == "Razer"].sort_values(by = ["Inches","Price_usd"], ascending = [False,True]) 
```
