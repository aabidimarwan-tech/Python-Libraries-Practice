### Multi-Level Indexing


```python
import pandas as pd
```

```python
df = pd.read_csv(r"C:\Users\seid\Downloads\creating-an-index-bleua.zip", encoding = "cp1252")

pd.set_option('display.max.rows', 1350)
pd.set_option('display.max.columns', 15)
df 
```

```python
df_continent = df.set_index(['continent', 'sub_region']).sort_values(['continent', 'sub_region', 'country'], ascending = [False, True, True])
```

```python
df_sub_region = df.set_index('sub_region').sort_values([ 'sub_region', 'country'], ascending = [True, True])
```

```python
df_sub_region.loc["Caribbean"]
```

```python
df_continent.loc['Asia']
```
