## Exercise 1: FizzBuzz
Print numbers 1 to 30.
- Multiples of 3 → print `Fizz`
- Multiples of 5 → print `Buzz`
- Multiples of both → print `FizzBuzz`

```python
def fizzbuzz():
    for number in range(1, 31):
        if number % 3 == 0 and number % 5 == 0:
            print("FizzBuzz")
        elif number % 3 == 0:
            print("Fizz")
        elif number % 5 == 0:
            print("Buzz")
        else:
            print(number)
    pass  # your code here
```

## Exercise 4: String practice

```python
text = "Data Analysis is fun and useful"

def count_words(s):
    """Count how many words are in the string."""
    word = s.split() 
    the_lengh = (len(word))
    return the_lengh 



def reverse_string(s):
    """Return the string reversed."""
    return s[::-1] 






def count_letter_a(s):
    """Count how many times 'a' appears, case-insensitive."""
    count = 0
    for i in s:
        if i.upper() == "A":
            count += 1
    return count 
```

```python
print(count_words(text))
print(reverse_string(text))
print(count_letter_a(text))
```
