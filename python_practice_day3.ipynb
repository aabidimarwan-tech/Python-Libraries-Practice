## Exercise 1: Loading a DataFrame
- `load_scores_df(filename)` → reads a `name,score` file into a pandas DataFrame with columns `["name", "score"]`. Make sure the `score` column is an integer type, not a string.

```python
import pandas as pd

def load_scores_df(filename):
    """Read a 'name,score' file into a DataFrame with columns ['name', 'score']."""
    stu_data = []
    with open(filename, "r") as f:
        for line in f:
            line = line.strip()
            if not line:
                continue
            name, score = line.split(",")
            stu_data.append((name, int(score)))
    Student_Data = pd.DataFrame(stu_data, columns = ["name", "score"])
    return Student_Data 
  
```

## Exercise 3: Filtering and adding columns
- `passing_students(df, threshold)` → returns only the rows where `score >= threshold`
- `add_grade_column(df)` → adds a new column `grade` to the DataFrame: `"A"` if score >= 90, `"B"` if >= 80, `"C"` if >= 70, else `"F"`

```python
def passing_students(df, threshold):
    """Return only the rows where score >= threshold."""
    
    passed_student = df[df["score"] >= threshold]

    return passed_student 

    pass  # your code here


def score_to_grade(score):
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    else:
        return "F"
        

def add_grade_column(df):
    """Add a 'grade' column: A >= 90, B >= 80, C >= 70, else F."""

    df["grade"] = df["score"].apply(score_to_grade) 
    return df 

```

```python
print(passing_students(df, 70))

df = add_grade_column(df)
print(df)
```

## Exercise 4: Grouping
- `count_by_grade(df)` → returns a Series (or DataFrame) showing how many students got each grade
- `average_by_grade(df)` → returns the average score for each grade group

```python
def count_by_grade(df):
    """Return the count of students per grade.""" 
    student_groups = df.groupby("grade")
    count_student = student_groups.size() 
    return count_student 
     


def average_by_grade(df):
    """Return the average score per grade."""
    student_groups = df.groupby("grade")
    average_grade = student_groups["score"].mean()  

    return average_grade 

    
```

```python
print(count_by_grade(df))
print(average_by_grade(df))
```
